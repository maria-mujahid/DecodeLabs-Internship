```python
Q1=db1['TotalPrice'].quantile(0.25)
Q3=db1['TotalPrice'].quantile(0.75)
IQR=Q3-Q1
lower_limit=Q1-1.5*IQR
upper_limit=Q3+1.5*IQR
print('IQR',IQR)
print('lower_limit',lower_limit)
print('upper_limit',upper_limit)
print('min TotalPrice',db1.TotalPrice.min())
print('max TotalPrice',db1.TotalPrice.max())
```

```python
sns.boxplot(x=db1['TotalPrice'])
plt.title('outliers detected for TotalPrice')
plt.show()
```

```python
# Recalculate the TotalPrice limits
Q1 = db1['TotalPrice'].quantile(0.25)
Q3 = db1['TotalPrice'].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Find potential outliers
price_outliers = db1[
    (db1['TotalPrice'] < lower_limit) |
    (db1['TotalPrice'] > upper_limit)
]

price_outliers[['OrderID', 'Product', 'Quantity', 'UnitPrice', 'TotalPrice']]
```

### Outlier Detection

The above cells show the outlier detection process for the `TotalPrice` column of the dataset. Based on the calculated lower limit of **-1341.4125** and upper limit of **3330.4075**, any total price below the lower limit or above the upper limit would be considered an outlier.

The minimum total price is **11.39**, while the maximum is **3456.40**. Since the maximum value exceeds the calculated upper limit, outliers are present in the `TotalPrice` column. The analysis identifies **eight outlier records**, all associated with orders having a quantity of 5. The highest outlier is a tablet order with a total price of **3456.40**.

These values are potential outliers according to the IQR method and do not necessarily indicate errors in the dataset.
